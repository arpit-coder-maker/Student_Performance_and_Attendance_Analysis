```python

# Student Performance and Attendance Analysis

import pandas as pd
import numpy as np



print("Following CLASSES Are Avalaible"
"CLASS10ASTUDENTS.csv"
" CLASS10BSTUDENTS.csv")
# Take file name from the user
file_name = input("Enter the CSV file name: ")

# Load the dataset
df = pd.read_csv(file_name)

# Take number of rows from the user
n = int(input("How many rows do you want to see? "))

# Print the rows
print("Here are the first", n, "rows:")
print(df.head(n))
```

```python
# Search Student Details by Roll Number

print("\n--- SEARCH STUDENT DETAILS ---")

roll_no = int(input("Enter Student Roll Number: "))

student = df[df["Roll No."] == roll_no]

if not student.empty:
    print("\nStudent Details Found:")
    print(student.to_string(index=False))
else:
    print("\nStudent with this Roll Number not found.")
```

```python
print("PERFORMANCE ANALYSIS")

# Top Scoring Students
print("\n--- TOP 5 SCORING STUDENTS ---")

top_students = df.sort_values(
    by="Percentage", ascending=False
).head(5)

print(top_students[["Roll No.", "Name", "Percentage"]])


# Low Scoring Students
print("\n--- LOWEST 5 SCORING STUDENTS ---")

low_students = df.sort_values(
    by="Percentage", ascending=True
).head(5)

print(low_students[["Roll No.", "Name", "Percentage"]])
```
